# Part 2 summary: mean and spread over seeds
Reads the official scorer output for each seed. Upload the `results/` folder from the repo first, or run after the two seed notebooks in the same session.

```python
import json, pandas as pd

RESULTS_DIR = "/content"
SEEDS = [13, 2026]

rows = []
for s in SEEDS:
    with open(f"{RESULTS_DIR}/seed{s}-scores.json") as f:
        sc = json.load(f)
    rows.append({"seed": s, "weighted_f1": sc["weighted_f1"],
                 "soft_cons": sc["soft_cons"], "hard_cons": sc["hard_cons"]})

scores = pd.DataFrame(rows)
print(scores.round(3).to_string(index=False))
```

```python
metrics = scores.drop(columns="seed")
summary = pd.DataFrame({"mean": metrics.mean(), "std": metrics.std(),   # sample std (ddof=1)
                        "min": metrics.min(), "max": metrics.max()})
print(summary.round(3))
```
